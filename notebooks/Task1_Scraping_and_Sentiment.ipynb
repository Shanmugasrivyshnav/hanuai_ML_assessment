## Sentiment
VADER is a baseline; manually validate sarcasm, multilingual and mixed reviews.

```python
path=PROJECT/"data"/"bestbuy_reviews.csv"
if path.exists():
 df=pd.read_csv(path).fillna({"title":"","review_text":""})
 vals=df.apply(lambda r: classify(str(r.title)+". "+str(r.review_text),r.rating),axis=1)
 df[["sentiment","sentiment_tag","sentiment_score"]]=pd.DataFrame(vals.tolist(),index=df.index)
 df.to_csv(path,index=False,encoding="utf-8-sig")
 display(df.head(),df.sentiment.value_counts())
else: print("Collect reviews first.")
```
